```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import urllib.request
from pathlib import Path
import cv2
```

```python
url="https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz"
urllib.request.urlretrieve(url,"mnist.npz")
```

```python
with np.load("mnist.npz") as f:
  x_train,y_train=f["x_train"], f["y_train"]
  x_test,y_test=f["x_test"], f["y_test"]
  print("training images", x_train.shape)
  print("training lables", y_train.shape)
  print("testing images", x_test.shape)
  print("testing lables",y_test.shape)

```

```python
img = x_train[10]

plt.imshow(img, cmap="gray")
plt.title("Original Image - Digit 3")
plt.axis("off")
plt.show()
```

```python
img = x_train[10]

plt.imshow(img, cmap="gray")
plt.title("Original Image - Digit 3")
plt.axis("off")
plt.show()
```

```python


x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0


print("Min:", x_train.min())
print("Max:", x_train.max())

```

```python
print("First 10 labels:")
print(y_train[:10])
```

```python
# Edge detection
img = (x_train[10] * 255).astype('uint8')
edges = cv2.Canny(img, 230, 255)

plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.show()
```

```python

img = x_train[10]
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

plt.imshow(rotated, cmap='gray')
plt.title("Rotated Image")
plt.show()
```

```python
img = x_train[10]
blur = cv2.blur(img, (3, 3))

plt.imshow(blur, cmap='gray')
plt.title("Normal Blur")
plt.show()
```

```python
img = x_train[10]
blur = cv2.GaussianBlur(img, (5, 5), 0)

plt.imshow(blur, cmap='gray')
plt.title("Gaussian Blur")
plt.show()
```

```python
img = x_train[10]
resized = cv2.resize(img, (10, 10))

plt.imshow(resized, cmap='gray')
plt.title("Resized Image (10x10)")
plt.show()
```
